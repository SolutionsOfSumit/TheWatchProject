# TheCountDownProject

The goal is to create a countdown which will show how much the time is passed

## Creating a basic timer

```python
import time 

# deadline_in_seconds = int(input("Type deadline in seconds"))
deadline_in_seconds = 10

while deadline_in_seconds >= 0:
    print(deadline_in_seconds)
    deadline_in_seconds -= 1
    time.sleep(1)

print("Times Up")
```

## Now the task is to make time display as 00:00:00

```python
import datetime as dt

hours = 0
minutes = 1
seconds = 10

while hours >= 0:
    while minutes >= 0:
        while seconds >= 0:
            deadline = dt.datetime(
                time.localtime().tm_year,
                time.localtime().tm_mon,
                time.localtime().tm_mday,
                hours,
                minutes,
                seconds
                ).timetuple()
            print(time.strftime('%H:%M:%S', deadline))
            seconds -= 1
            time.sleep(1)
        minutes -= 1
        seconds = 59
    hours -= 1
    minutes = 59

print("Time's Up")
```

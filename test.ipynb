# TheCountDownProject

The goal is to create a countdown which will show how much the time is passed

## Creating a basic timer

```python
import time 

# deadline_in_seconds = int(input("Type deadline in seconds"))
deadline_in_seconds = 10

while deadline_in_seconds >= 0:
    print(deadline_in_seconds)
    deadline_in_seconds -= 1
    time.sleep(1)

print("Times Up")
```

## Now the task is to make time display as 00:00:00

```python
import datetime as dt

hours = 0
minutes = 1
seconds = 2

while hours >= 0:
    while minutes >= 0:
        while seconds >= 0:
            deadline = dt.datetime(
                time.localtime().tm_year,
                time.localtime().tm_mon,
                time.localtime().tm_mday,
                hours,
                minutes,
                seconds
                ).timetuple()
            print(time.strftime('%H:%M:%S', deadline))
            seconds -= 1
            # time.sleep(1)
        minutes -= 1
        seconds = 59
    hours -= 1
    minutes = 59

print("Time's Up")
```

# Adding user inputs

```python
import time
import datetime as dt

hours = int(input("Numbers of hours to setup: "))
while hours < 0:
    print("Hours can not be negative")
    hours = int(input("Numbers of hours to setup: "))

minutes = int(input("Numbers of Minutes to setup: "))
while minutes > 59 or minutes < 0:
    print("Enter minutes withing the range of 59")
    minutes = int(input("Numbers of Minutes to setup: "))

seconds = int(input("Number of Seconds to setup: "))
while seconds > 59 or seconds < 0:
    print("Enter seconds withing the range of 59")
    seconds = int(input("Numbers of seconds to setup: "))

if hours == 0 and minutes == 0 and seconds == 0:
    print("There is no time to count down")
else:
    while hours >= 0:
        while minutes >= 0:
            while seconds >= 0:
                deadline = dt.datetime(
                    time.localtime().tm_year,
                    time.localtime().tm_mon,
                    time.localtime().tm_mday,
                    hours,
                    minutes,
                    seconds
                    ).timetuple()
                print(time.strftime('%H:%M:%S', deadline))
                seconds -= 1
                time.sleep(1)
            minutes -= 1
            seconds = 59
        hours -= 1
        minutes = 59

    print("Time's Up")
```

# Detecting keyboard input while running code in console.

Aperently we have some modules we can work with one of them is keyboard by python itself



```python
import keyboard 
import time

count_down = 10
is_paused = False
while count_down >= 0:
    if keyboard.is_pressed('p') and is_paused == False:
        print("You have stoped countdown")
        count_down = count_down
        is_paused == True

    print(count_down)
    count_down -= 1
    time.sleep(1)
```

```python
import keyboard 
import time

count_down = 10

while count_down >= 0:
    if keyboard.is_pressed('p'):
        input("Press any key to continue")

    if keyboard.is_pressed('s'):
        print(f"You stopped the countdown at {count_down}")
        break

    print(count_down)
    count_down -= 1
    time.sleep(1)
```

# Second Solution 

This time we took help from chatGPT for logic because it was not pitch perfect.

```python
import keyboard
import time

tik = 10
is_paused = False

print("Press P to pause/resume and S to stop")

while tik >= 0:

    if keyboard.is_pressed('s'):
        print(f"Countdown stopped at {tik}")
        break

    if keyboard.is_pressed('p'):
        is_paused = not is_paused

        if is_paused:
            print("Paused")
        else:
            print("Resumed")

        keyboard.wait('p')  # Wait until P is released

    if not is_paused:
        print(tik)
        tik -= 1
        time.sleep(1)

print("Countdown finished")
```

# Adding restart to the project

```python
import keyboard
import time

tik = 10
restart = tik

is_paused = False

print("Press P to pause/resume and S to stop")

while tik >= 0:

    if keyboard.is_pressed('s'):
        print(f"Countdown stopped at {tik}")
        break
    if keyboard.is_pressed('r'):
        tik = restart

    if keyboard.is_pressed('p'):
        is_paused = not is_paused

        if is_paused:
            print("Paused")
        else:
            print("Resumed")

        keyboard.wait('p')  # Wait until P is released

    if not is_paused:
        print(tik)
        tik -= 1
        time.sleep(1)


print("Countdown finished")
```

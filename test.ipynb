The goal is to create a countdown which will show how much the time is passed

## Creating a basic timer

```python
import time 

# deadline_in_seconds = int(input("Type deadline in seconds"))
deadline_in_seconds = 10

while deadline_in_seconds >= 0:
    print(deadline_in_seconds)
    deadline_in_seconds -= 1
    time.sleep(1)

print("Times Up")
```

## Now the task is to make time display as 00:00:00

```python
import datetime as dt

hours = 0
minutes = 1
seconds = 2

while hours >= 0:
    while minutes >= 0:
        while seconds >= 0:
            deadline = dt.datetime(
                time.localtime().tm_year,
                time.localtime().tm_mon,
                time.localtime().tm_mday,
                hours,
                minutes,
                seconds
                ).timetuple()
            print(time.strftime('%H:%M:%S', deadline))
            seconds -= 1
            # time.sleep(1)
        minutes -= 1
        seconds = 59
    hours -= 1
    minutes = 59

print("Time's Up")
```

However, there are still three things to address:
1. Pause: When paused, your loop continuously checks the keyboard. That's okay functionally, but it wastes CPU cycles.
2. Stop: Pressing S exits the inner loop, but the outer loop immediately starts another iteration. You need a stopped state that waits for R.
3. Restart: Pressing R inside the countdown resets the timer, but it doesn't prevent the current iteration from decrementing it immediately.

```python
import keyboard
import time

tik = 10
restart = tik

is_paused = False
is_quit = False

print("Press P to pause/resume and S to stop")

while is_quit != True:

    if keyboard.is_pressed('r'):
        print("Restarting the countdown")
        tik = restart
        is_paused = False

    while tik >= 0:

        if keyboard.is_pressed('q'):
            print(f"program quit")
            is_quit = True
            break

        if keyboard.is_pressed('s'):
            print(f"Countdown stopped at {tik}")
            break

        if keyboard.is_pressed('r'):
            print("Restarting the countdown")
            tik = restart
            is_paused = False
        

        if keyboard.is_pressed('p'):
            is_paused = not is_paused

            if is_paused:
                print("Paused")
            else:
                print("Resumed")

            keyboard.wait('p')  

        if not is_paused:
            print(tik)
            tik -= 1
            time.sleep(1)

    keyboard.wait('r')  # Issue is here

print("Countdown finished")
```

## Validating using pydantic

```python
import time
import keyboard
import datetime as dt
from pydantic import BaseModel, ValidationError, AfterValidator
from typing import Annotated

# Schema functions

def is_hour(value: int):
    if not 0 <= value <= 23:
        raise ValidationError("Value is not an valid")
    return value

def is_min(value: int):
    if not 0 <= value <= 59:
        raise ValidationError("Value is not an valid")
    return value

# Schemas

class DeadLine(BaseModel):
    hour : Annotated[int, AfterValidator(is_hour)]  = time.localtime().tm_hour
    minute : Annotated[int, AfterValidator(is_min)]  = time.localtime().tm_min
    second : Annotated[int, AfterValidator(is_min)]  = time.localtime().tm_sec

    

# Functions 

def get_deadline(hour, minutes, seconds):
    user_data = {
        "hour" : hour,
        "minute" : minutes,
        "second" : seconds 
    }

    deadline = DeadLine(**user_data).model_dump()

    return deadline

print(get_deadline(21,21,50))
```
